```python
import numpy as np
import matplotlib.pyplot as plt
import time
import random

# Turn on interactive mode
plt.ion()

# Create a figure and axis
fig, ax = plt.subplots()

# Initial data setup: 1 row with N angles (e.g., 500)
num_angles = 500
gradient_row = np.zeros((1, num_angles))

# Create the image plot
im = ax.imshow(gradient_row, cmap='gray', aspect='auto', extent=[0, num_angles, 0, 1])
ax.set_title("Distance Gradient Over Angle")
ax.get_yaxis().set_visible(False)

# Dynamic update loop
for _ in range(100):  # simulate 100 frames
    # Simulate a new distance scan row (0–5m mapped to grayscale 0–255)
    new_row = np.array([random.uniform(0, 5) for _ in range(num_angles)])
    gradient_row[0] = new_row

    # Update the image data
    im.set_data(gradient_row)
    im.set_clim(0, 5)  # optional: fix color range for contrast
    fig.canvas.draw()
    fig.canvas.flush_events()
    time.sleep(0.1)

plt.ioff()
plt.show()
```
